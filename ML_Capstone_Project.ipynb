## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Merge ratings and books dataframes
df = pd.merge(ratings, books, on='ISBN')
df.head()
````

````python
df = pd.merge(df, users, on='User-ID', how='left')
# Drop unnecessary columns
df.drop(['Image-URL-S', 'Image-URL-M', 'Image-URL-L'], axis=1, inplace=True)
df.head()
````

````python
# Check for missing values
df.isnull().sum()
````

````python
# Drop rows with missing values
df.dropna(inplace=True)

# Check for missing values again
df.isnull().sum()
````

````python
# Convert 'Year-Of-Publication' to numeric, coercing errors
df['Year-Of-Publication'] = pd.to_numeric(df['Year-Of-Publication'], errors='coerce')

# Replace 0 with NaN (assuming 0 represents missing or invalid data)
df['Year-Of-Publication'] = df['Year-Of-Publication'].replace(0, np.nan)

# Filter out years that are clearly in the future or before a reasonable publication year
current_year = pd.to_datetime('now').year
df = df[(df['Year-Of-Publication'] >= 1800) & (df['Year-Of-Publication'] <= current_year)].copy()

# Drop rows with remaining NaN values in 'Year-Of-Publication'
df.dropna(subset=['Year-Of-Publication'], inplace=True)

# Convert 'Year-Of-Publication' to integer after cleaning
df['Year-Of-Publication'] = df['Year-Of-Publication'].astype(int)
````

Data Wrangling and Manipulations Performed:

Column Name Cleaning:

Applied clean_column_names to books, users, and ratings DataFrames. This standardizes column names to snake_case (e.g., Book-Title became book_title), ensuring consistency and easier programmatic access.

Applied convert_types to all three DataFrames. This function attempts to convert columns to more appropriate data types (e.g., object to numeric where applicable), which can improve memory efficiency and facilitate numerical operations.

Identified Invalid Entries: The year_of_publication column contained non-numeric values (e.g., 'DK Publishing Inc', 'Gallimard') and unrealistic year values (like 0 or years far in the future). These were initially coerced to NaN during numeric conversion.
Imputation and Outlier Treatment: Rows with year_of_publication as 0 or greater than 2024 (assuming current year is 2024, future years are outliers) were replaced with NaN. The missing values were then imputed using the median year_of_publication (which is robust to outliers) to maintain data integrity without discarding records.
Type Conversion: After cleaning and imputation, the column was converted to int type, as years are whole numbers.
Handling Missing age in users:

Identified Invalid Entries: The age column had NaN values and potentially unrealistic age values (like 0 or extremely high ages > 100). These were initially coerced to NaN during numeric conversion.
Imputation and Outlier Treatment: Rows with age as 0 or greater than 100 were replaced with NaN. The missing age values were then imputed using the median age from the remaining valid entries. This is a common and robust strategy for handling missing numerical data, especially when the distribution might be skewed or contain outliers.
Type Conversion: The age column was converted to int type after imputation.
Insights Found:

Improved Data Quality: These steps have significantly improved the quality of the books and users DataFrames by addressing structural issues (column names), type inconsistencies, and critical missing/invalid data points in key numerical columns.
Readiness for Analysis: The year_of_publication and age columns are now in a clean, numeric format, making them suitable for statistical analysis, visualization, and use in machine learning models.

##### 2. What is/are the insight(s) found from the chart?

From the distribution of book ratings, we can observe the following key insights:

Dominance of Implicit Ratings (Rating 0): A very large proportion of ratings are '0'. In this dataset, a rating of '0' often indicates an implicit rating, meaning the user read the book but did not explicitly provide a numerical rating. This suggests that a significant portion of user interactions are not explicitly rated.
Skew Towards Higher Explicit Ratings: Among explicit ratings (1-10), there's a general trend where higher ratings (e.g., 7, 8, 10) are more frequent than lower ratings (e.g., 1, 2, 3). This indicates that users who bother to give an explicit rating tend to rate books positively.
Relatively Fewer Low Explicit Ratings: Ratings of 1, 2, and 3 are much less common compared to higher explicit ratings, implying that truly disliked books are either rated implicitly as '0' or are simply not rated explicitly with low scores as often as liked books are rated highly.
Sparse Distribution for Explicit Ratings: The counts for explicit ratings, while present, are significantly lower than the count for implicit ratings. This highlights the challenge of working with explicit feedback in this dataset.

**Will the gained insights help creating a positive business impact?**

Yes, the gained insights can help create a positive business impact, particularly for a book recommendation system or an online bookstore.

*   **Understanding User Behavior:** Knowing the distribution of ratings, especially the high frequency of implicit ratings (0), highlights the importance of considering implicit feedback when building recommendation systems. Explicit ratings (1-10) are also valuable, and the skew towards higher ratings for explicit feedback suggests that users who actively rate tend to rate favorably.
*   **Targeted Marketing and Recommendations:** Identifying popular rating ranges (7, 8, 10) can inform marketing strategies and help in recommending books that are likely to receive higher explicit ratings.
*   **Identifying Potential Issues:** While not necessarily indicating negative growth in itself, the high number of implicit ratings (0) could suggest that users are not always prompted or encouraged to provide explicit feedback. A business could investigate ways to encourage more explicit ratings to gain richer data.

**Are there any insights that lead to negative growth? Justify with specific reason.**

The insights from this specific chart (distribution of book ratings) *by themselves* do not directly lead to or indicate negative growth. The presence of a large number of implicit ratings (0) is a characteristic of the dataset and how ratings were collected, rather than an indicator of negative user sentiment or business decline.

However, *if* a business solely relied on explicit ratings and ignored the vast amount of implicit feedback (rating 0), it could lead to:

*   **Inaccurate Recommendations:** The recommendation system would miss a large portion of user interactions, potentially leading to less relevant suggestions.
*   **Missed Opportunities:** The business might not fully understand which books are being consumed (even without explicit positive ratings), missing opportunities for engagement or targeted promotions based on reading history.

Therefore, while the insight itself isn't negative, *ignoring* it could negatively impact the business's ability to effectively understand and engage with its users.

Null Hypothesis (H0): The true average explicit book rating is 7.5.\nAlternative Hypothesis (H1): The true average explicit book rating is not 7.5.

### Hypothetical Statement - 2

Formulate a null and alternative hypothesis concerning the relationship between book ratings and publication years (e.g., comparing average publication years for highly-rated vs. lower-rated books). Perform an appropriate statistical test (e.g., independent samples t-test). Explain the chosen test and the conclusion drawn from the p-value.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.


**Null Hypothesis (H0)**: There is no significant difference in the average publication year between highly-rated books (Book-Rating >= 8) and lower-rated books (Book-Rating < 8).

**Alternative Hypothesis (H1)**: There is a significant difference in the average publication year between highly-rated books (Book-Rating >= 8) and lower-rated books (Book-Rating < 8).

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import ttest_ind

# Filter for explicit ratings only
explicit_ratings_df_filtered = df[df['Book-Rating'] > 0].copy()

# Define highly-rated and lower-rated books
highly_rated_books = explicit_ratings_df_filtered[explicit_ratings_df_filtered['Book-Rating'] >= 8]['Year-Of-Publication']
lower_rated_books = explicit_ratings_df_filtered[explicit_ratings_df_filtered['Book-Rating'] < 8]['Year-Of-Publication']

# Perform independent samples t-test
t_statistic, p_value = ttest_ind(highly_rated_books, lower_rated_books, equal_var=False) # Assuming unequal variances

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
````

##### Which statistical test have you done to obtain P-Value?

Independent samples t-test.

##### Why did you choose the specific statistical test?

The independent samples t-test was chosen because we are comparing the means of two independent groups: 'Year-Of-Publication' for highly-rated books and 'Year-Of-Publication' for lower-rated books. The goal is to determine if there is a statistically significant difference between these two means. This test is appropriate when comparing the means of two distinct groups, especially when assuming unequal variances (as indicated by equal_var=False in the code), which is often a safer assumption in real-world data.

#### Conclusion from the P-value

With a P-value of 0.3522 (which is greater than the common significance level of 0.05), we **fail to reject the null hypothesis**. This means there is no statistically significant difference in the average publication year between highly-rated books (Book-Rating >= 8) and lower-rated books (Book-Rating < 8). In simpler terms, based on this dataset and test, older books are not significantly more or less likely to receive high ratings compared to newer books, or vice-versa, when considering explicit ratings.


Now that the outliers have been visualized, the next step is to quantify them using the Interquartile Range (IQR) method. Calculating Q1, Q3, and then the upper and lower bounds will allow for precise identification of outliers.



````python
Q1 = users['Age'].quantile(0.25)
Q3 = users['Age'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): {Q1}")
print(f"Q3 (75th percentile): {Q3}")
print(f"IQR (Interquartile Range): {IQR}")
print(f"Lower Bound for Outliers: {lower_bound}")
print(f"Upper Bound for Outliers: {upper_bound}")
````

````python
users_treated = users.copy() # Make a copy to avoid SettingWithCopyWarning

# Cap values below the lower bound
users_treated['Age'] = users_treated['Age'].apply(lambda x: lower_bound if x < lower_bound else x)

# Cap values above the upper bound
users_treated['Age'] = users_treated['Age'].apply(lambda x: upper_bound if x > upper_bound else x)

print("Outliers in 'Age' column treated using capping method.")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

The outlier treatment technique used for the 'Age' column in the `users` DataFrame is **Capping with Interquartile Range (IQR)**.

**Rationale for choosing this method:**

1.  **Identification of Outliers:** The IQR method is a robust statistical approach for identifying outliers. It defines outliers as data points that fall below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR`, where Q1 is the 25th percentile, Q3 is the 75th percentile, and IQR is the interquartile range (Q3 - Q1). This method is less sensitive to extreme values than methods based on the mean and standard deviation.

2.  **Preservation of Data:** Instead of entirely removing rows containing outliers (which can lead to significant data loss, especially with a high number of outliers like in the 'Age' column), capping replaces the extreme values with the calculated upper or lower bounds. This approach retains the data points within the dataset, preventing potential loss of valuable information that might be associated with these users.

3.  **Maintenance of Distribution Shape (to some extent):** While it modifies the extreme ends of the distribution, capping aims to pull in the outliers without drastically altering the overall shape of the central distribution. This is beneficial for many statistical analyses and machine learning models that might be sensitive to extreme values but still rely on the general distribution of the data.

4.  **Practicality for Age Data:** For 'Age', values that are extremely high (e.g., above 100) or extremely low (e.g., single digits that don't fit typical user behavior for a book rating platform) are often data entry errors or represent unusual cases that could skew models. Capping them to reasonable limits (like 44 for the upper bound and 20 for the lower bound in this specific dataset's IQR calculation) makes the data more usable without discarding the user entirely.

In summary, capping with IQR was chosen because it effectively handles extreme values in a numerical feature like 'Age' by standardizing them to a plausible range, thereby improving data quality for subsequent analysis and modeling without incurring significant data loss.

#### 2. Lemmatization


*    Lemmatization is the process of reducing words to their base or dictionary form, known as a lemma. Unlike stemming, which often chops off word endings and can result in non-words, lemmatization uses vocabulary and morphological analysis to return the root word. For example, \"running\", \"ran\", \"runs\" would all be reduced to \"run\". This is applied to the tokenized lists for 'Book-Title', 'Book-Author', and 'Publisher'.

    *    By transforming different inflected forms of a word into a single base form, lemmatization helps reduce the dimensionality of the text data. This is particularly useful for features like 'Book-Title' and 'Book-Author' where variations in word forms (e.g., plurals, verb tenses) might occur.
    *    It helps in grouping words with similar meanings, ensuring that different forms of a word are treated as the same concept. This improves the accuracy of text analysis, search, and recommendation systems by ensuring that, for example, a search for \"run\" also captures documents containing \"running\" or \"ran\".
    *    Lemmatization often produces valid words that are semantically meaningful, unlike stemming which can produce truncated words that are not recognized. This preserves more information and accuracy in the processed text.
    *    In datasets where textual features might be sparse (i.e., many unique words but low frequency for each), reducing words to their lemmas helps in consolidating entries and increasing the effective frequency of terms, leading to more robust models.

````python
import nltk

# Download 'averaged_perceptron_tagger_eng' if not already downloaded
try:
    nltk.data.find('taggers/averaged_perceptron_tagger_eng')
except LookupError:
    nltk.download('averaged_perceptron_tagger_eng')
    print("NLTK 'averaged_perceptron_tagger_eng' downloaded.")

def pos_tag_tokens(tokens):
    # Ensure tokens is a list. If it was a string representation of a list, it should have been converted earlier.
    # After tokenization and lemmatization, each element in df[col] should be a list of strings.
    if not isinstance(tokens, list):
        # This case should ideally not happen if prior steps are correct and df[col] consistently holds lists.
        # If it's a string, try to tokenize it.
        if isinstance(tokens, str):
            tokens = nltk.word_tokenize(tokens)
        else:
            return [] # Unhandled type, return empty list

    # Now 'tokens' is guaranteed to be a list.
    # Check if the list elements are already tuples (meaning it's already POS tagged).
    if tokens and isinstance(tokens[0], tuple):
        # If elements are tuples (e.g., [('word', 'TAG')]), extract just the words.
        cleaned_tokens = [word for word, tag in tokens]
    else:
        # Otherwise, assume it's a list of strings (the expected input for fresh POS tagging).
        cleaned_tokens = tokens

    return nltk.pos_tag(cleaned_tokens)

columns_to_process = ['Book-Title', 'Book-Author', 'Publisher']

for col in columns_to_process:
    # Apply the POS tagging function to each column
    df[col] = df[col].apply(lambda x: pos_tag_tokens(x))

print("POS tagging applied to 'Book-Title', 'Book-Author', and 'Publisher' columns.")

# Display sample entries from modified columns to verify
print("\nSample entries after POS tagging:")
for col in columns_to_process:
    print(f"\n--- {col} ---")
    print(df[col].head().tolist())
````

````python
# Install gensim library if not already installed
!pip install gensim

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from gensim.models import Word2Vec

# --- TF-IDF Vectorization ---

# Step 1: Join the lemmatized tokens back into strings for TF-IDF
# TfidfVectorizer expects string input, so join the list of tokens with a space.
# Ensure the columns exist and are in the expected list format from previous preprocessing.

# Safely convert to list of strings if not already
def join_tokens(tokens):
    if isinstance(tokens, list) and all(isinstance(t, tuple) for t in tokens):
        # If elements are (word, tag) tuples, extract only the word
        return ' '.join([word for word, tag in tokens])
    elif isinstance(tokens, list):
        # If elements are just strings (e.g., if POS tagging was skipped or resulted differently)
        return ' '.join(tokens)
    return str(tokens) # Handle cases where it might not be a list

df['Book-Title_TFIDF_Input'] = df['Book-Title'].apply(join_tokens)
df['Book-Author_TFIDF_Input'] = df['Book-Author'].apply(join_tokens)
df['Publisher_TFIDF_Input'] = df['Publisher'].apply(join_tokens)

print("Prepared text columns for TF-IDF.")

# Step 2: Initialize and apply TF-IDF Vectorizers
# Using max_features to limit the vocabulary size, which is good practice.

tfidf_vectorizer_title = TfidfVectorizer(max_features=5000) # Example: limit to 5000 most frequent terms
tfidf_matrix_title = tfidf_vectorizer_title.fit_transform(df['Book-Title_TFIDF_Input'])

tfidf_vectorizer_author = TfidfVectorizer(max_features=2000) # Example: limit to 2000 terms
tfidf_matrix_author = tfidf_vectorizer_author.fit_transform(df['Book-Author_TFIDF_Input'])

tfidf_vectorizer_publisher = TfidfVectorizer(max_features=1000) # Example: limit to 1000 terms
tfidf_matrix_publisher = tfidf_vectorizer_publisher.fit_transform(df['Publisher_TFIDF_Input'])

print("\nTF-IDF Vectorization complete.")
print(f"Shape of Book-Title TF-IDF matrix: {tfidf_matrix_title.shape}")
print(f"Shape of Book-Author TF-IDF matrix: {tfidf_matrix_author.shape}")
print(f"Shape of Publisher TF-IDF matrix: {tfidf_matrix_publisher.shape}")


# --- Word2Vec Vectorization ---

# Step 3: Train Word2Vec models on the tokenized text columns
# The 'Book-Title', 'Book-Author', 'Publisher' columns should already contain lists of tokens from previous steps.

# Ensure the input to Word2Vec is a list of lists of strings (words only, not tuples)
# Filter out empty lists or non-list entries if any for robustness
def extract_words_from_pos_tags(tokens):
    if isinstance(tokens, list) and all(isinstance(t, tuple) for t in tokens):
        return [word for word, tag in tokens]
    return tokens # Return as is if not a list of tuples

valid_title_sentences = [extract_words_from_pos_tags(tokens) for tokens in df['Book-Title'].tolist() if isinstance(tokens, list) and tokens]
valid_author_sentences = [extract_words_from_pos_tags(tokens) for tokens in df['Book-Author'].tolist() if isinstance(tokens, list) and tokens]
valid_publisher_sentences = [extract_words_from_pos_tags(tokens) for tokens in df['Publisher'].tolist() if isinstance(tokens, list) and tokens]

# Train Word2Vec model for Book-Titles
word2vec_model_title = Word2Vec(sentences=valid_title_sentences, vector_size=100, window=5, min_count=1, workers=4)
print("\nWord2Vec model trained for Book-Titles.")

# Train Word2Vec model for Book-Authors
word2vec_model_author = Word2Vec(sentences=valid_author_sentences, vector_size=50, window=3, min_count=1, workers=4)
print("Word2Vec model trained for Book-Authors.")

# Train Word2Vec model for Publishers
word2vec_model_publisher = Word2Vec(sentences=valid_publisher_sentences, vector_size=50, window=3, min_count=1, workers=4)
print("Word2Vec model trained for Publishers.")

# Helper function to get a document vector by averaging word vectors
def get_document_vector(model, tokens):
    # Filter out words not in the vocabulary of the trained model
    valid_tokens = [token for token in tokens if token in model.wv]
    if not valid_tokens:
        # Return a zero vector of the model's vector_size if no words are in vocab
        return [0.0] * model.vector_size
    # Average the word vectors to get the document vector
    return sum(model.wv[token] for token in valid_tokens) / len(valid_tokens)

# Example: Apply the function to create a new column for book title vectors
df['Book-Title_W2V_Vector'] = df['Book-Title'].apply(lambda x: get_document_vector(word2vec_model_title, extract_words_from_pos_tags(x)))
print("\nAdded Word2Vec vectors for Book-Titles to DataFrame.")
print(f"Shape of a single Book-Title Word2Vec vector: {len(df['Book-Title_W2V_Vector'].iloc[0])}")
````

````python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from gensim.models import Word2Vec
from sklearn.preprocessing import LabelEncoder

# --- Re-initialize df from scratch for robustness ---
# These steps are repeated from earlier cells to ensure df is defined
from google.colab import drive
drive.mount('/content/drive')# Mount your Google Drive to the '/content/drive' directory in Colab.

books = pd.read_csv('/content/drive/MyDrive/Almabetter/Books.csv')
users = pd.read_csv('/content/drive/MyDrive/Almabetter/Users.csv')
ratings = pd.read_csv('/content/drive/MyDrive/Almabetter/Ratings.csv')

# Merge ratings and books dataframes
df = pd.merge(ratings, books, on='ISBN')

# Drop unnecessary columns (Image-URL-S, Image-URL-M, Image-URL-L)
df.drop(['Image-URL-S', 'Image-URL-M', 'Image-URL-L'], axis=1, inplace=True)

# Drop rows with missing values in 'Book-Author' or 'Publisher'
df.dropna(subset=['Book-Author', 'Publisher'], inplace=True)

# Convert 'Year-Of-Publication' to numeric and clean
df['Year-Of-Publication'] = pd.to_numeric(df['Year-Of-Publication'], errors='coerce')
df['Year-Of-Publication'] = df['Year-Of-Publication'].replace(0, np.nan)
current_year = pd.to_datetime('now').year
df = df[(df['Year-Of-Publication'] >= 1800) & (df['Year-Of-Publication'] <= current_year)].copy()
df.dropna(subset=['Year-Of-Publication'], inplace=True)
df['Year-Of-Publication'] = df['Year-Of-Publication'].astype(int)

# Handle missing and outlier 'Age' in 'users' and merge to df
users['Age'].fillna(users['Age'].median(), inplace=True)
Q1 = users['Age'].quantile(0.25)
Q3 = users['Age'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
users['Age'] = users['Age'].apply(lambda x: lower_bound if x < lower_bound else x)
users['Age'] = users['Age'].apply(lambda x: upper_bound if x > upper_bound else x)

# Merge cleaned age into df
df = pd.merge(df, users[['User-ID', 'Age']], on='User-ID', how='left')

# Apply Label Encoding (copied from earlier cell OzU_uaPdeFTY)
le = LabelEncoder()
categorical_cols_to_encode = ['Book-Title', 'Book-Author', 'Publisher']
for col in categorical_cols_to_encode:
    if col in df.columns and df[col].dtype == 'object':
        df[col + '_Encoded'] = le.fit_transform(df[col])
    else:
        print(f"Warning: Column '{col}' not found or not of object dtype in df. Skipping encoding.")

# Apply Text Preprocessing (copied from earlier cells for robustness)
import re
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Download NLTK data if not already downloaded
try: nltk.data.find('corpora/stopwords')
except LookupError: nltk.download('stopwords')
try: nltk.data.find('tokenizers/punkt')
except LookupError: nltk.download('punkt')
try: nltk.data.find('corpora/wordnet')
except LookupError: nltk.download('wordnet')
try: nltk.data.find('corpora/omw-1.4')
except LookupError: nltk.download('omw-1.4')
try: nltk.data.find('taggers/averaged_perceptron_tagger_eng')
except LookupError: nltk.download('averaged_perceptron_tagger_eng')

CONTRACTION_MAP = {"ain't": "am not", "aren't": "are not", "can't": "cannot", "can't've": "cannot have", "'cause": "because", "could've": "could have", "couldn't": "could not", "couldn't've": "could not have", "didn't": "did not", "doesn't": "does not", "don't": "do not", "hadn't": "had not", "hadn't've": "had not have", "hasn't": "has not", "haven't": "have not", "he'd": "he would", "he'd've": "he would have", "he'll": "he will", "he'll've": "he will have", "he's": "he is", "how'd": "how did", "how'd'y": "how do you", "how'll": "how will", "how's": "how is", "I'd": "I would", "I'd've": "I would have", "I'll": "I will", "I'll've": "I will have", "I'm": "I am", "I've": "I have", "isn't": "is not", "it'd": "it would", "it'd've": "it would have", "it'll": "it will", "it'll've": "it will have", "it's": "it is", "let's": "let us", "ma'am": "madam", "mayn't": "may not", "might've": "might have", "mightn't": "might not", "mightn't've": "might not have", "must've": "must have", "mustn't": "must not", "mustn't've": "must not have", "needn't": "need not", "needn't've": "need not have", "o'clock": "of the clock", "oughtn't": "ought not", "oughtn't've": "ought not have", "shan't": "shall not", "sha'n't": "shall not", "shan't've": "shall not have", "she'd": "she would", "she'd've": "she would have", "she'll": "she will", "she'll've": "she will have", "she's": "she is", "should've": "should have", "shouldn't": "should not", "shouldn't've": "should not have", "so've": "so have", "so's": "so is", "that'd": "that would", "that'd've": "that would have", "that's": "that is", "there'd": "there would", "there'd've": "there would have", "there's": "there is", "these's": "these is", "they'd": "they would", "they'd've": "they would have", "they'll": "they will", "they'll've": "they will have", "they're": "they are", "they've": "they have", "to've": "to have", "wasn't": "was not", "we'd": "we would", "we'd've": "we would have", "we'll": "we will", "we'll've": "we will have", "we're": "we are", "we've": "we have", "weren't": "were not", "what'll": "what will", "what'll've": "what will have", "what're": "what are", "what's": "what is", "what've": "what have", "when's": "when is", "when've": "when have", "where'd": "where did", "where's": "where is", "where've": "where have", "who'll": "who will", "who'll've": "who will have", "who's": "who is", "who've": "who have", "why's": "why is", "why've": "why have", "will've": "will have", "won't": "will not", "won't've": "will not have", "would've": "would have", "wouldn't": "would not", "wouldn't've": "would not have", "y'all": "you all", "y'all'd": "you all would", "y'all'd've": "you all would have", "y'all're": "you all are", "y'all've": "you all have", "you'd": "you would", "you'd've": "you would have", "you'll": "you will", "you'll've": "you will have", "you're": "you are", "you've": "you have"}

def expand_contractions(text):
    if not isinstance(text, str):
        return text
    text = text.lower()
    for contraction, expansion in CONTRACTION_MAP.items():
        text = re.sub(r'\b' + contraction + r'\b', expansion, text)
    return text

def remove_punctuations(text):
    if not isinstance(text, str):
        return text
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)

def remove_urls_and_digits_from_words(text):
    if not isinstance(text, str):
        return text
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'\b\w*\d\w*\b', '', text)
    return text

english_stopwords = set(stopwords.words('english'))
def remove_stopwords(text):
    if not isinstance(text, str):
        return text
    return ' '.join([word for word in text.split() if word not in english_stopwords])

def remove_extra_whitespaces(text):
    if not isinstance(text, str):
        return text
    return re.sub(r'\s+', ' ', text).strip()

def rephrase_text(text):
    if not isinstance(text, str):
        return text
    rephrase_map = { # No specific rules defined in previous cell, but keeping structure
    }
    for old, new in rephrase_map.items():
        text = text.replace(old, new)
    return text

lemmatizer = WordNetLemmatizer()
def lemmatize_tokens(tokens):
    if not isinstance(tokens, list):
        return tokens
    return [lemmatizer.lemmatize(token) for token in tokens]

def pos_tag_tokens(tokens):
    if not isinstance(tokens, list):
        if isinstance(tokens, str):
            tokens = word_tokenize(tokens)
        else:
            return []
    if tokens and isinstance(tokens[0], tuple):
        cleaned_tokens = [word for word, tag in tokens]
    else:
        cleaned_tokens = tokens
    return nltk.pos_tag(cleaned_tokens)

columns_to_process_text = ['Book-Title', 'Book-Author', 'Publisher']

for col in columns_to_process_text:
    df[col] = df[col].astype(str).apply(expand_contractions)
    df[col] = df[col].astype(str).str.lower()
    df[col] = df[col].astype(str).apply(remove_punctuations)
    df[col] = df[col].astype(str).apply(remove_urls_and_digits_from_words)
    df[col] = df[col].astype(str).apply(remove_stopwords)
    df[col] = df[col].astype(str).apply(remove_extra_whitespaces)
    df[col] = df[col].astype(str).apply(rephrase_text)
    df[col] = df[col].apply(word_tokenize) # Tokenization
    df[col] = df[col].apply(lemmatize_tokens) # Lemmatization
    df[col] = df[col].apply(pos_tag_tokens) # POS Tagging


# --- Downsample the DataFrame to manage memory ---
sample_size = 10000  # Limiting sample size for robust execution, changed from 50000

if len(df) > sample_size:
    df = df.sample(n=sample_size, random_state=42).reset_index(drop=True)
    print(f"DataFrame downsampled to {len(df)} rows.")
else:
    print("DataFrame size is already smaller than or equal to the sample_size. No downsampling performed.")
print(f"Current DataFrame shape: {df.shape}\n")

# --- Helper functions (re-defined for self-containment) ---
def extract_words_from_pos_tags(tokens):
    if not isinstance(tokens, list):
        if isinstance(tokens, str):
            return tokens.split() # Try to split if it's a string that wasn't properly tokenized
        return [] # Return empty list for other non-list types
    if not tokens: # Handle empty list
        return []
    if isinstance(tokens[0], tuple): # If elements are (word, tag) tuples
        return [w for w, _ in tokens]
    else: # If elements are just strings
        return tokens

def join_tokens(tokens):
    if isinstance(tokens, list) and all(isinstance(t, tuple) for t in tokens):
        return ' '.join([word for word, tag in tokens])
    elif isinstance(tokens, list):
        return ' '.join(tokens)
    return str(tokens)

# Helper for Word2Vec document vector
def get_document_vector(model, tokens, default_size):
    tokens = extract_words_from_pos_tags(tokens) # Ensure tokens are just words
    if model is None or not hasattr(model, 'wv') or not model.wv.key_to_index:
        return np.zeros(default_size)
    valid = [t for t in tokens if t in model.wv.key_to_index]
    if not valid:
        return np.zeros(model.vector_size)
    return np.mean(model.wv[valid], axis=0)


# --- Prepare text columns for TF-IDF and Word2Vec (after potential downsampling) ---
df['Book-Title_TFIDF_Input'] = df['Book-Title'].apply(join_tokens)
df['Book-Author_TFIDF_Input'] = df['Book-Author'].apply(join_tokens)
df['Publisher_TFIDF_Input'] = df['Publisher'].apply(join_tokens)
print("Prepared text columns for TF-IDF and Word2Vec.")


# --- TF-IDF Vectorization ---
tfidf_vectorizer_title = TfidfVectorizer(max_features=5000)
tfidf_matrix_title = tfidf_vectorizer_title.fit_transform(df['Book-Title_TFIDF_Input'])

tfidf_vectorizer_author = TfidfVectorizer(max_features=2000)
tfidf_matrix_author = tfidf_vectorizer_author.fit_transform(df['Book-Author_TFIDF_Input'])

tfidf_vectorizer_publisher = TfidfVectorizer(max_features=1000)
tfidf_matrix_publisher = tfidf_vectorizer_publisher.fit_transform(df['Publisher_TFIDF_Input'])

print("TF-IDF Vectorization complete.")

# --- Apply TruncatedSVD for dimensionality reduction on TF-IDF features ---

# For Book-Title TF-IDF
svd_title = TruncatedSVD(n_components=200, random_state=42)
tfidf_svd_title = svd_title.fit_transform(tfidf_matrix_title)
tfidf_df_title = pd.DataFrame(tfidf_svd_title, columns=[f'title_svd_{i}' for i in range(200)])
print(f"Shape of TF-IDF SVD title DataFrame: {tfidf_df_title.shape}")

# For Book-Author TF-IDF
svd_author = TruncatedSVD(n_components=100, random_state=42)
tfidf_svd_author = svd_author.fit_transform(tfidf_matrix_author)
tfidf_df_author = pd.DataFrame(tfidf_svd_author, columns=[f'author_svd_{i}' for i in range(100)])
print(f"Shape of TF-IDF SVD author DataFrame: {tfidf_df_author.shape}")

# For Publisher TF-IDF
svd_publisher = TruncatedSVD(n_components=50, random_state=42)
tfidf_svd_publisher = svd_publisher.fit_transform(tfidf_matrix_publisher)
tfidf_df_publisher = pd.DataFrame(tfidf_svd_publisher, columns=[f'publisher_svd_{i}' for i in range(50)])
print(f"Shape of TF-IDF SVD publisher DataFrame: {tfidf_df_publisher.shape}")


# --- Word2Vec Model Training and Vector Generation ---
def train_w2v_robust(col_series, size):
    sents = [extract_words_from_pos_tags(x) for x in col_series.tolist() if isinstance(x, list) and x]
    if not sents:
        print(f"Warning: No valid sentences found for column '{col_series.name}'. Returning a dummy Word2Vec model with empty vocabulary.")
        model = Word2Vec(vector_size=size, min_count=1)
        model.build_vocab([["dummyword"]])
        model.wv.key_to_index = {}
        return model
    else:
        print(f"Training Word2Vec for column '{col_series.name}' with {len(sents)} sentences.")
        model = Word2Vec(sentences=sents, vector_size=size, min_count=1, workers=4)
        return model

word2vec_model_title = train_w2v_robust(df['Book-Title'], size=100)
print("Word2Vec model processed for Book-Titles.")

word2vec_model_author = train_w2v_robust(df['Book-Author'], size=50)
print("Word2Vec model processed for Book-Authors.")

word2vec_model_publisher = train_w2v_robust(df['Publisher'], size=50)
print("Word2Vec model processed for Publishers.")

# Create Word2Vec vectors for all columns and add to df
df['Book-Title_W2V_Vector'] = df['Book-Title'].apply(lambda x: get_document_vector(word2vec_model_title, x, 100))
df['Book-Author_W2V_Vector'] = df['Book-Author'].apply(lambda x: get_document_vector(word2vec_model_author, x, 50))
df['Publisher_W2V_Vector'] = df['Publisher'].apply(lambda x: get_document_vector(word2vec_model_publisher, x, 50))


# --- Expand Word2Vec vectors into separate columns ---
w2v_df_title = pd.DataFrame(df['Book-Title_W2V_Vector'].tolist(), columns=[f'title_w2v_{i}' for i in range(100)])
print(f"Shape of W2V title DataFrame: {w2v_df_title.shape}")

w2v_df_author = pd.DataFrame(df['Book-Author_W2V_Vector'].tolist(), columns=[f'author_w2v_{i}' for i in range(50)])
print(f"Shape of W2V author DataFrame: {w2v_df_author.shape}")

w2v_df_publisher = pd.DataFrame(df['Publisher_W2V_Vector'].tolist(), columns=[f'publisher_w2v_{i}' for i in range(50)])
print(f"Shape of W2V publisher DataFrame: {w2v_df_publisher.shape}")


# --- Concatenate all features into a single DataFrame ---
# Ensure encoded columns exist in df after downsampling
required_encoded_cols = ['Book-Title_Encoded', 'Book-Author_Encoded', 'Publisher_Encoded']
for col in required_encoded_cols:
    if col not in df.columns:
        print(f"Warning: Encoded column '{col}' not found in df. Attempting to re-encode.")
        original_col_name = col.replace('_Encoded', '')
        le = LabelEncoder()
        df[col] = le.fit_transform(df[original_col_name])

combined_features_df = pd.concat([
    df[['User-ID', 'Book-Rating', 'Year-Of-Publication', 'Age']],
    df[required_encoded_cols],
    tfidf_df_title,
    tfidf_df_author,
    tfidf_df_publisher,
    w2v_df_title,
    w2v_df_author,
    w2v_df_publisher
], axis=1)

# Handle NaN values that might appear during concatenation (e.g., from W2V if a row had no valid words)
combined_features_df.fillna(0, inplace=True)

# Display the first 5 rows
print("\nFirst 5 rows of the combined feature DataFrame:")
print(combined_features_df.head())
````

#### 2. Feature Selection

````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import SelectFromModel
import pandas as pd

# 1. Separate features (X) and target variable (y)
X = combined_features_df.drop('Book-Rating', axis=1)
y = combined_features_df['Book-Rating']

print(f"Original number of features: {X.shape[1]}")

# 2. Initialize a RandomForestRegressor model
# Using a small n_estimators for speed, a higher number might be used for better feature importance calculation
estimator = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)

# 3. Instantiate SelectFromModel using the RandomForestRegressor and fit it to X and y
sf = SelectFromModel(estimator, threshold='median') # Select features whose importance is greater than the median
sf.fit(X, y)

# 4. Transform the original features (X) to retain only the selected features
X_selected = sf.transform(X)

# Get the mask of selected features
selected_feature_mask = sf.get_support()

# 5. Get the names of the selected features
selected_feature_names = X.columns[selected_feature_mask].tolist()

# 6. Print the number of features before and after selection, and then print the list of selected features
print(f"Number of features after selection: {X_selected.shape[1]}")
print("\nSelected Features:")
for feature in selected_feature_names:
    print(feature)
````
